### 1.2. Построение графика

```python
!pip uninstall matplotlib -y
!pip install matplotlib
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file = "TRB_10_7.tsv"
data = pd.read_csv(file, sep='\t')

# Подсчёт частот использования генов TRBV
trbv_counts = data['TRBV7-9'].value_counts()  # Используем правильное имя столбца

# Построение графика
plt.figure(figsize=(10, 6)) 
sns.barplot(x=trbv_counts.index, y=trbv_counts.values, palette="viridis")  # Столбчатая диаграмма

plt.title("Частота использования генов TRBV", fontsize=16)  
plt.xlabel("Гены TRBV", fontsize=14)  
plt.ylabel("Частота", fontsize=14)  
plt.xticks(rotation=45, ha='right')  
plt.grid(axis='y', linestyle='--', alpha=0.7) 
plt.tight_layout() 
plt.show()
```
