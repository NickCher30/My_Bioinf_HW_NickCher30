## Краткая справка о выбранных мною белках:

**Цитохром c / Cytochrome c** — небольшой гемсодержащий белок, который участвует в переносе электронов в дыхательной цепи митохондрий, связывая комплексы III (цитохром bc1) и IV (цитохром c оксидаза), то есть, это электрон-переносящий белок, участвующий в осуществлении клеточного дыхания. Состоит из $\approx$ 100 аминокислот и содержит гемовую группу, которая обратимо изменяет степень окисления железа, что и позволяет ему переносить электроны. Цитохром c высококонсервативен у разных видов. Также цитохром с участвует в процессах апоптоза (программируемой клеточной смерти), высвобождаясь из митохондрий в цитоплазму. 

- P99999 (Homo sapiens) — цитохром c человека.


- P62897 (Mus musculus) — цитохром c мыши.


- P00004 (Equus caballus) — цитохром c лошади.


- P00044 (Saccharomyces cerevisiae) — цитохром c дрожжей (точнее - его изоформа 1).
